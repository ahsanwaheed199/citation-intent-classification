```python
import json
import os

# Check the scicite folder structure
scicite_root = "../scicite"

print("=" * 70)
print("SCICITE FOLDER STRUCTURE")
print("=" * 70)

# List all folders and files in scicite
for root, dirs, files in os.walk(scicite_root):
    level = root.replace(scicite_root, '').count(os.sep)
    indent = ' ' * 2 * level
    print(f'{indent}{os.path.basename(root)}/')
    
    subindent = ' ' * 2 * (level + 1)
    for file in files[:5]:  # Show first 5 files per folder
        print(f'{subindent}{file}')
    
    if len(files) > 5:
        print(f'{subindent}... and {len(files) - 5} more files')
    
    if level > 2:  # Stop going too deep
        break
```

```python
import urllib.request
import tarfile
import os

print("Downloading SciCite data...")
print("This may take 1-2 minutes...")

url = "https://s3-us-west-2.amazonaws.com/ai2-s2-research/scicite/scicite.tar.gz"
filename = "../scicite_data.tar.gz"

try:
    # Download
    urllib.request.urlretrieve(url, filename)
    print(f"✓ Downloaded: {filename}")
    
    # Extract
    print("Extracting files...")
    with tarfile.open(filename, 'r:gz') as tar:
        tar.extractall(path="../")
    
    print("✓ Extracted!")
    
    # Check what we got
    data_path = "../scicite/data"
    if os.path.exists(data_path):
        files = os.listdir(data_path)
        print(f"\nFiles in data folder: {files}")
    
except Exception as e:
    print(f"Error: {e}")
```

```python
import shutil
import os

# The tar extracted to: ../scicite/scicite/
# We want the data at: ../scicite/data/

source = "../scicite/scicite"
dest = "../scicite/data"

if os.path.exists(source):
    print(f"Moving {source} to {dest}")
    
    # Remove dest if it exists
    if os.path.exists(dest):
        shutil.rmtree(dest)
    
    # Move
    shutil.move(source, dest)
    
    print("✓ Done!")
    
    # Verify
    files = os.listdir(dest)
    print(f"\nFiles in scicite/data: {files}")
else:
    print(f"Source not found: {source}")
    
    # Check what's in scicite folder
    scicite_contents = os.listdir("../scicite")
    print(f"Contents of ../scicite: {scicite_contents}")
```

```python
import json

# Count the 3 citation intent classes
label_counts = {}
filepath = "../scicite/train.jsonl"

with open(filepath, 'r', encoding='utf-8') as f:
    for line in f:
        example = json.loads(line)
        label = example['label']
        label_counts[label] = label_counts.get(label, 0) + 1

print("=" * 70)
print("SCICITE LABEL DISTRIBUTION (Training Data)")
print("=" * 70)

total = sum(label_counts.values())
for label, count in sorted(label_counts.items(), key=lambda x: x[1], reverse=True):
    percentage = (count / total) * 100
    print(f"{label:20s}: {count:6d} ({percentage:5.1f}%)")

print(f"{'TOTAL':20s}: {total:6d}")

# Show examples of each class
print("\n" + "=" * 70)
print("EXAMPLE FROM EACH CLASS")
print("=" * 70)

examples_by_label = {}
with open(filepath, 'r', encoding='utf-8') as f:
    for line in f:
        example = json.loads(line)
        label = example['label']
        if label not in examples_by_label:
            examples_by_label[label] = example

for label in sorted(examples_by_label.keys()):
    ex = examples_by_label[label]
    print(f"\n[{label.upper()}]")
    print(f"Sentence: {ex['string'][:120]}...")
    print(f"Section: {ex['sectionName']}")
```
