```python
import pandas as pd

# Load the SOFT dataset
df = pd.read_csv("../soft/dataset/ACLARC_SOFT.tsv", sep="\t")

print("Shape:", df.shape)
print("\nThe label we'll use (aclarc_label):")
print(df["aclarc_label"].value_counts())

print("\nOne example citation_context:")
print(df["citation_context"].iloc[0])
```

```python
from sklearn.model_selection import train_test_split

# Keep just the two columns we need: text + label
data = df[["citation_context", "aclarc_label"]].copy()
data = data.rename(columns={"citation_context": "text", "aclarc_label": "label"})

# Drop any rows with missing text or label (safety)
data = data.dropna(subset=["text", "label"])

# Split: 80% train, 20% test.
# stratify=label keeps the same class proportions in both sets.
train_df, test_df = train_test_split(
    data,
    test_size=0.2,
    stratify=data["label"],
    random_state=42,
)

print(f"Train: {len(train_df)} examples")
print(f"Test:  {len(test_df)} examples")
print(f"\nTrain label balance:\n{train_df['label'].value_counts()}")
```
