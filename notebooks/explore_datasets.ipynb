```python
import json
import os
```

```python
# Check what folders we have 
folders = ['../acl-arc', '../scicite', '../soft']

for folder in folders:
    exists = os.path.exists(folder)
    print(f"{folder}: {'✓ Found' if exists else '✗ Not found'}")
```

```python
# Let's see what's actually in these files
acl_arc_path = "../acl-arc/data/adjudicated-and-supplemental"

# Pick a few different files and check their size
json_files = [f for f in os.listdir(acl_arc_path) if f.endswith('.json')]

print(f"Total JSON files: {len(json_files)}\n")

# Check first 5 files
for file in json_files[:5]:
    filepath = f"{acl_arc_path}/{file}"
    size = os.path.getsize(filepath)
    print(f"{file}: {size} bytes")
```

```python
# Read a file with actual data
filepath = "../acl-arc/data/adjudicated-and-supplemental/A00-1004.json"

try:
    with open(filepath, 'r') as f:
        example = json.load(f)
    
    print("=" * 70)
    print("ONE ACL-ARC EXAMPLE")
    print("=" * 70)
    
    print(f"\nType: {type(example)}")
    print(f"\nKeys: {example.keys()}")
    
    # Show first 1000 characters
    print(f"\nFirst part of data:")
    print(json.dumps(example, indent=2)[:1500])
    
except Exception as e:
    print(f"Error: {e}")
    # If JSON fails, try reading as text
    with open(filepath, 'r') as f:
        content = f.read()
    print(f"File size: {len(content)} chars")
    print(f"First 500 chars:\n{content[:500]}")
```

```python
# Check if there's a simpler data format
import os

acl_arc_root = "../acl-arc"

# List all files in root
all_files = []
for root, dirs, files in os.walk(acl_arc_root):
    for file in files:
        filepath = os.path.join(root, file)
        # Only show interesting files
        if file.endswith(('.csv', '.tsv', '.jsonl', '.txt')) and 'README' not in file:
            all_files.append(filepath)

print("Simpler format files found:")
for f in all_files[:20]:
    print(f)
```

```python
# Read one .txt file to see the format
filepath = "../acl-arc/data/adjudicated-and-supplemental/A00-1004-parscit.130908.xml.txt"

with open(filepath, 'r') as f:
    content = f.read()

print("=" * 70)
print("ACL-ARC .txt FILE FORMAT")
print("=" * 70)
print(f"\nFile size: {len(content)} characters")
print(f"\nFirst 1000 characters:\n")
print(content[:1000])
```

```python
# Read one .ann file (annotation file with labels)
filepath = "../acl-arc/data/adjudicated-and-supplemental/A00-1004-parscit.130908.xml.ann"

with open(filepath, 'r', encoding='utf-8') as f:
    content = f.read()

print("=" * 70)
print("ACL-ARC .ann FILE FORMAT (ANNOTATIONS)")
print("=" * 70)
print(f"\nFirst 2000 characters:\n")
print(content[:2000])
```
